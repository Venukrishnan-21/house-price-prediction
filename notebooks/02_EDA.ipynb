# Univariate Analysis 
(One Variable at a Time)

**Price Distribution**

```python
plt.figure(figsize=(15,10))
sns.histplot(ds['price_realistic'], bins=50, kde=True)
plt.title("Distribution of House Prices")
plt.xlabel("Price")
plt.ylabel("House Count")

plt.savefig("house_price_distribution_1.png", dpi=300, bbox_inches="tight")
plt.show()
```

**Outlier Detection (Boxplot)**

```python
plt.figure(figsize=(8,8))
sns.boxplot(x=ds['price_realistic'])
plt.title("Price Outliers")

plt.savefig("house_price_distribution_2.png", dpi=300, bbox_inches="tight")
plt.show()
```

**Price vs Area**

```python
plt.figure(figsize=(6,5))
sns.scatterplot(x='area', y='price_realistic', data=ds)
plt.title("Area vs Price")

plt.savefig("house_price_distribution_3.png", dpi=300, bbox_inches="tight")
plt.show()
```

**Location vs Price**

```python
plt.figure(figsize=(6,4))
sns.boxplot(x='location', y='price_realistic', data=ds)
plt.title("Price by Location")

plt.savefig("house_price_distribution_4.png", dpi=300, bbox_inches="tight")
plt.show()
```

**Income Level vs Price**

```python
plt.figure(figsize=(6,4))
sns.boxplot(x='income_level', y='price_realistic', data=ds)
plt.title("Price by Income Level")

plt.savefig("house_price_distribution_5.png", dpi=300, bbox_inches="tight")
plt.show()
```

**Amenity Count vs Price**

```python
ds['amenity_count'] = (
    ds['garage'] + ds['parking'] + ds['garden'] + ds['security']
)
```

```python
plt.figure(figsize=(6,4))
sns.boxplot(x='amenity_count', y='price_realistic', data=ds)
plt.title("Amenities vs Price")

plt.savefig("house_price_distribution_6.png", dpi=300, bbox_inches="tight")
plt.show()
```

**Distance & Crime Effect**

```python
sns.scatterplot(x='distance', y='price_realistic', data=ds)
plt.title("Distance vs Price")

plt.savefig("house_price_distribution_7.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
sns.scatterplot(x='crime_rate', y='price_realistic', data=ds)
plt.title("Crime Rate vs Price")

plt.savefig("house_price_distribution_8.png", dpi=300, bbox_inches="tight")
plt.show()
```

**Correlation Heatmap**

```python
plt.figure(figsize=(12,8))
sns.heatmap(
    ds.select_dtypes(include=np.number).corr(),
    cmap='coolwarm',
    annot=False
)
plt.title("Feature Correlation Heatmap")

plt.savefig("house_price_distribution_9.png", dpi=300, bbox_inches="tight")
plt.show()
```
