# Realistic Price Formula
(Indian Context)

```python
base_price_per_sqft = {
    'low': 3000,
    'medium': 6000,
    'premium': 9000
}

ds['base_price'] = ds['area'] * ds['location'].map(base_price_per_sqft)

# Amenities bonus
ds['amenities_score'] = (
    ds['garage'] + ds['parking'] + ds['garden'] + ds['security']
) * 100000  # ₹1 lakh each

# Distance penalty
ds['distance_penalty'] = ds['distance'] * 20000

# Crime penalty
ds['crime_penalty'] = ds['crime_rate'] * 50000

# Final realistic price
ds['price_realistic'] = (
    ds['base_price']
    + ds['amenities_score']
    - ds['distance_penalty']
    - ds['crime_penalty']
)

ds['price_realistic'] = ds['price_realistic'].clip(lower=500000)
```

```python
ds.head()
```

While validating synthetic data, I identified unrealistic pricing where large premium houses were undervalued. I redesigned the price-generation logic using real-world per-square-foot pricing, location multipliers, and amenity premiums to align the dataset with actual Indian real estate markets.

| Location | ₹ per sq.ft | Why                      |
| -------- | ----------- | ------------------------ |
| low      | 3,000       | Outskirts, less demand   |
| medium   | 6,000       | Regular city areas       |
| premium  | 9,000       | Posh / high-demand areas |


| Amenity  | Value     |
| -------- | --------- |
| Garage   | ₹1,00,000 |
| Parking  | ₹1,00,000 |
| Garden   | ₹1,00,000 |
| Security | ₹1,00,000 |


###  Distance penalty

ds['distance_penalty'] = ds['distance'] * 20000

Distance = 10 km

Penalty = 10 × 20,000 = ₹2,00,000

###  crime penalty

ds['crime_penalty'] = ds['crime_rate'] * 50000

Crime Rate = 2.5

Penalty = 2.5 × 50,000 = ₹1,25,000

### Final realistic price

Final Price =(Base Area Price)+ (Amenities Bonus)− (Distance Penalty)− (Crime Penalty)


Base Price        = ₹1,08,00,000
Amenities Bonus   = ₹3,00,000
Distance Penalty  = ₹2,00,000
Crime Penalty     = ₹1,25,000

Final Price = ₹1,07,75,000 (~1.08 Cr)


### Minimum Price Safety Rule

It prevents house prices from becoming unrealistically low.

clip(lower=500000) means:

If the house price is less than ₹5,00,000, set it to ₹5,00,000

clip() is a pandas function used to limit values within a range.

Think of it as a price guard.
